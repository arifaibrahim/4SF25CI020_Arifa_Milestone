```python
df['rate'] = df['rate'].fillna(df['rate'].mode()[0])
df
```

```python
df['rate'] = df['rate'].replace('NEW', np.nan)
df['rate'] = df['rate'].replace('-', np.nan)
df['rate'] = df['rate'].astype(str).str.replace('/5', '', regex=False)
df['rate'] = pd.to_numeric(df['rate'], errors='coerce')
```

```python
def rating_category(rating):
    if pd.isna(rating):
        return 'not rated'
    elif rating >=4:
        return 'Excelllent'
    elif rating >=3:
        return 'average'
    else:
        return 'poor'
df['rating_category']= df['rate'].apply(rating_category)
```

```python
top_locs = df['location'].value_counts().head(10)
plt.figure(figsize=(12, 6))
sns.barplot(
    x=top_locs.values,
    y=top_locs.index
)
plt.title('Top 10 Locations by num of Restaurants')
plt.xlabel('No. of restaurants')
plt.ylabel('Location')

plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['rate'].dropna(), bins=20, kde=True)
plt.title('Distribution of ratings')
plt.xlabel('rating')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(9, 6))
sns.boxplot(
    x='online_order',
    y='rate',
    data=df
)
plt.title('Restaurant ratings based on online orders')
plt.xlabel('Online orrdering')
plt.ylabel('Rating')
plt.show()
```
